# Regresión Lineal


<p style='text-align: justify;'>Este modelo nos permite encontrar la relación que existe entre una variable dependiente y una independiente, así como predecir el valor esperado de una variable aleatoria y cuando x toma un valor especifico.</p>  
<p style='text-align: justify;'>Este método implica suponer que alguna función de la recta de forma y=mx+b es la representación mas adecuada de la distribución de los datos en un diagrama de dispersión. En un mundo ideal, todos los puntos del diagrama de dispersión se alinearían en una sola línea recta perfecta sin la necesidad de encontrar la recta que resume los datos ya que solo bastaría con unir dichos puntos, sin embargo, la realidad no es así, típicamente en los conjuntos de datos de los diagramas de dispersión se puede trazar mas de una línea recta y cada una de estas líneas dará un ajuste diferente al conjunto de datos, se trata entonces de encontrar la línea recta que brinde el mayor ajuste.</p>

```python
# Preparamos datos
X= df['LSTAT'].values.reshape(-1,1)
y = df["MEDV"].values.reshape(-1,1)
```

```python
X.shape
```

```python
#Creamos un split de datos
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.33, random_state=100)
```

```python
from sklearn.linear_model import LinearRegression
#Creando el modelo y entrenando
reg = LinearRegression().fit(X_train,y_train)
#Prediciendo valores de entrenamiento
y_train_hat = reg.predict(X_train)
#Prediciendo valores de validación
y_test_hat = reg.predict(X_test)
```

```python
import matplotlib.pyplot as plt
import numpy as np
# Le decimos a jupyter que grafique en  el cuaderno
%matplotlib inline
# Creamos un scatter plot con los datos de entrenamiento
plt.scatter(X_train, y_train)
# Creamos un scatter plot con los datos de validación
plt.scatter(X_test, y_test)
# En X_plot guardamos valores distribuidos entre 0 y 40
X_plot = np.linspace(0,40).reshape(-1, 1)
# Con el modelo predecimos X_plot
y_plot = reg.predict(X_plot)
# Graficamos el modelo
plt.plot(X_plot, y_plot,"r--")
```

### Análisis de residuos

Una vez ajustado el modelo lineal y representada la recta de regresión sobre los datos, es importante comprobar qué tan bien se adapta el modelo.  
Para ello se utilizan los **residuos**, que son las diferencias entre los valores reales y los valores predichos por el modelo:

$$ r_i = y_i - \hat{y}_i $$

- **Gráfico de residuos vs. predicciones**:  
  Muestra los errores en función de lo que predice el modelo.  
  Si el modelo lineal es adecuado, los puntos deberían distribuirse de forma aleatoria alrededor de cero (sin formar curvas ni patrones).  

- **Histograma de residuos**:  
  Permite ver cómo se distribuyen los errores.  
  Lo esperado es que la mayoría de residuos estén cerca de cero y que la distribución sea aproximadamente simétrica.

Estos gráficos indican si el modelo lineal describe razonablemente bien los datos o si, por el contrario, hay estructura en los errores que el modelo no logra capturar.


```python
import matplotlib.pyplot as plt

# Cálculo de residuos en el conjunto de test
residuos = y_test - y_test_hat

# Gráfico de residuos vs. predicciones
plt.scatter(y_test_hat, residuos, alpha=0.7)
plt.axhline(y=0, color='red', linestyle='--')
plt.xlabel("Valores predichos ($\hat{y}$)")
plt.ylabel("Residuos (y - ŷ)")
plt.title("Gráfico de residuos")
plt.show()

# Histograma de los residuos
plt.hist(residuos, bins=20, edgecolor='black', alpha=0.7)
plt.xlabel("Residuo")
plt.ylabel("Frecuencia")
plt.title("Distribución de los residuos")
plt.show()
```

### Métricas de desempeño

Calculamos el coeficiente de determinación $R^2$ tanto en entrenamiento como en prueba.  
Este valor indica qué fracción de la variabilidad de los datos logra explicar el modelo (cuanto más cercano a 1, mejor ajuste).


```python
# Calculamos el  error
from sklearn.metrics import r2_score
print("Entrenamiento", r2_score(y_train, y_train_hat))
print("Prueba", r2_score(y_test, y_test_hat))
```

## Regresión múltiple

En la regresión lineal simple buscamos predecir una variable de respuesta a partir de una única variable independiente.  
Es natural preguntarse qué ocurre si utilizamos **múltiples variables independientes** en el mismo modelo: esto es la regresión lineal múltiple.  
Con este ejemplo veremos cómo cambia la calidad del ajuste al incluir más información en el modelo.


```python
# Preparación de datos iniciales
X = df[["RM","AGE"]]
#podemos usar todo el conjunto, excepto la columna MEDV que usamos para y
# X=df.drop("MEDV", axis=1)
#tambien podemos quitar variabls con poca correlacion como CHAS
# X=df.drop("MEDV","CHAS", axis=1)
y = df["MEDV"].values.reshape(-1, 1)

# Hacemos un split de 33%
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=100)

# Se entrena el modelo
reg = LinearRegression().fit(X_train,y_train)

# Hacemos las  predicciones
y_train_hat = reg.predict(X_train)
y_test_hat = reg.predict(X_test)

# Calculamos el error
print("Entrenamiento", r2_score(y_train, y_train_hat))
print("Prueba", r2_score(y_test, y_test_hat))
```

### Reflexión sobre los resultados

Observamos que el $R^2$ en entrenamiento aumentó, pero en validación disminuyó.  
Esto indica que agregar variables no siempre mejora el desempeño del modelo en datos nuevos.  

Podemos probar con una variable más significativa, como `LSTAT`, o incluso con todo el conjunto de variables.  

- La **regresión lineal múltiple** es una extensión directa de la regresión lineal simple.  
- Aunque es un modelo muy sencillo de aplicar, la **selección de variables** es fundamental: elegir mal las variables puede empeorar la capacidad predictiva del modelo.
