```python
import numpy as np 
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
```

Acerca del conjunto de datos
Se han recopilado datos p sobre un conjunto de pacientes, todos con la misma enfermedad. Durante su tratamiento, cada paciente respondió a uno de los 5 medicamentos, A, B, C, X e Y. 

Se pretende construir un modelo para averiguar qué medicamento podría ser apropiado para un futuro paciente con la misma enfermedad. Los conjuntos de características de este conjunto de datos son Edad, Sexo, Presión arterial y Colesterol de los pacientes, y el objetivo es el medicamento al que cada paciente respondió. 

Es una muestra de clasificador binario, y puede usar la parte de entrenamiento del conjunto de datos para construir un árbol de decisión y, a continuación, usarlo para predecir la clase de un paciente desconocido, o para prescribirlo a un nuevo paciente. 


```python
my_data = pd.read_csv("drug200.csv", delimiter=",")
my_data[0:5]
```

<div href="pre-processing">
    <h2>Pre-processing</h2>
</div>

Usando my_data como los datos de Drug.csv leídos por los pandas, seleccionamos las variables:<br/>
X como matriz de características <br/>
Y como vector de respuesta  

```python
X = my_data[['Age', 'Sex', 'BP', 'Cholesterol', 'Na_to_K']].values
X[0:5]
```

Algunas características de este conjunto de datos son categóricas, como Sexo o BP. Los árboles de decisión de Sklearn no manejan variables categóricas

```python
from sklearn import preprocessing
le_sex = preprocessing.LabelEncoder()
le_sex.fit(['F','M'])
X[:,1] = le_sex.transform(X[:,1]) 


le_BP = preprocessing.LabelEncoder()
le_BP.fit([ 'LOW', 'NORMAL', 'HIGH'])
X[:,2] = le_BP.transform(X[:,2])


le_Chol = preprocessing.LabelEncoder()
le_Chol.fit([ 'NORMAL', 'HIGH'])
X[:,3] = le_Chol.transform(X[:,3]) 

X[0:5]
```

Ahora introducimos la variable de destino.

```python
y = my_data["Drug"]
y[0:5]
```

<hr>

<div id="setting_up_tree">
    <h2>Configuración del árbol de decisiones</h2>
   
</div>

```python
from sklearn.model_selection import train_test_split
```

```python
X_trainset, X_testset, y_trainset, y_testset = train_test_split(X, y, test_size=0.3, random_state=3)
```

<hr>

<div id="modelado">
    <h2>Modelado</h2>
    Primero crearemos una instancia <b>DecisionTreeClassifier</b> llamada <b>drugTree</b>.<br>
    Dentro del clasificador, especificamos el criterio "entropía" para que podamos ver la ganancia de información de cada nodo. 
    
</div>

```python
drugTree = DecisionTreeClassifier(criterion="entropy", max_depth = 4)
drugTree 
```

```python
drugTree.fit(X_trainset,y_trainset)
```

## Predicciones
   

```python
predTree = drugTree.predict(X_testset)
```

```python
print (predTree[0:5])
print (y_testset[0:5])
```

<hr>

<div id="evaluation">
    <h2>Evaluación</h2>
</div>

```python
from sklearn import metrics
import matplotlib.pyplot as plt
print("DecisionTrees's Accuracy: ", metrics.accuracy_score(y_testset, predTree))
```

__Accuracy clasification score__ calcula la precisión del subconjunto: el conjunto de etiquetas predichas para una muestra debe coincidir exactamente con el conjunto correspondiente de etiquetas en y_true.  

En la clasificación multietiqueta, la función devuelve la precisión del subconjunto. Si todo el conjunto de etiquetas predichas para una muestra coincide estrictamente con el conjunto verdadero de etiquetas, la precisión del subconjunto es 1.0; de lo contrario es 0.0.
